```python
# Pick controller 0 and print its capabilities.
js = pygame.joystick.Joystick(0)
js.init()

print(f"Name:    {js.get_name()}")
print(f"GUID:    {js.get_guid()}")
print(f"Buttons: {js.get_numbuttons()}")
print(f"Axes:    {js.get_numaxes()}")
print(f"Hats:    {js.get_numhats()}")
```

## Live watcher

Run the cell below, then press buttons / move sticks / use the D-pad. It prints each
input **as it changes** along with its index. Note the numbers you care about, then
**interrupt the kernel** (the stop ■ button) to end the loop.

```python
import time

DEADZONE = 0.5  # only report stick/trigger movement past this

prev_buttons = [js.get_button(i) for i in range(js.get_numbuttons())]
prev_hats = [js.get_hat(i) for i in range(js.get_numhats())]
prev_axes = [0.0] * js.get_numaxes()

print("Watching... press inputs on the controller. Interrupt the kernel to stop.\n")
try:
    while True:
        pygame.event.pump()

        for i in range(js.get_numbuttons()):
            v = js.get_button(i)
            if v != prev_buttons[i]:
                print(f"button {i:>2}  -> {'PRESSED' if v else 'released'}")
                prev_buttons[i] = v

        for i in range(js.get_numhats()):
            v = js.get_hat(i)
            if v != prev_hats[i]:
                print(f"hat    {i:>2}  -> {v}")
                prev_hats[i] = v

        for i in range(js.get_numaxes()):
            v = js.get_axis(i)
            if abs(v - prev_axes[i]) > 0.3 and abs(v) > DEADZONE:
                print(f"axis   {i:>2}  = {v:+.2f}")
                prev_axes[i] = v

        time.sleep(1 / 60)
except KeyboardInterrupt:
    print("\nStopped watching.")
```

## Reference: standard XInput layout

Most XInput pads (incl. the Cyclone 2 in XInput mode) follow this. Confirm against the
live watcher — your unit may differ.

| Index | Button | | Axis | Control |
|---|---|---|---|---|
| 0 | A | | 0 | Left stick X |
| 1 | B | | 1 | Left stick Y |
| 2 | X | | 2 | Right stick X |
| 3 | Y | | 3 | Right stick Y |
| 4 | Left bumper (LB) | | 4 | Left trigger (LT) |
| 5 | Right bumper (RB) | | 5 | Right trigger (RT) |
| 6 | Back/View | | | |
| 7 | Start/Menu | | | |
| 8 | Left stick click | | | |
| 9 | Right stick click | | | |

The **D-pad** is usually `hat 0`: left=`(-1,0)`, right=`(1,0)`, up=`(0,1)`, down=`(0,-1)`.

Put the numbers you found into `config.json` (`buttons`, `bumpers`, `hat`, `axes`). Valid
action strings are listed in the README.
